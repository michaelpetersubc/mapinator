```julia
using HTTP, JSON, PrettyTables, JLD, MySQL, DotEnv, DataFrames, LinearAlgebra
cfg = DotEnv.config("../../.env")
files_path = cfg["files_path"]
```

```julia
placement_rates = load(files_path*"placement_rates.jld")["placement_rates"]
```

```julia
n2 = placement_rates ./ sum(placement_rates, dims=2)
```

```julia
p2 = placement_rates[1:end, 2:end]
```

```julia
p2p = p2 ./ sum(p2,dims=2)
```

```julia
p2r = reverse(p2p, dims=2)
```

```julia
 normp2 = cumsum(p2r,dims=2)
```

```julia
normalized = new_matrix ./ sum(new_matrix, dims=2)
```
